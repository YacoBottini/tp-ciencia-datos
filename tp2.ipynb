Si abriéramos este archivo manualmente en Excel o lo miráramos crudo, notaríamos un par de cosas peligrosas para Python:

Separadores engañosos: Hay columnas (como la descripción o el elenco) que contienen textos largos llenos de comas. Si Python cortara a ciegas por cada coma, rompería la tabla entera. Pandas soluciona esto respetando las comillas dobles (") que envuelven esos textos.

Caracteres especiales (Encoding): Al ser un catálogo internacional, hay nombres con tildes y símbolos raros. Si lo abrimos con la codificación por defecto, Python podría tirar un UnicodeDecodeError. Vamos a tener que forzar la lectura pasándole un parámetro de encoding (como latin-1).

Columnas fantasma: Cada fila tiene un montón de comas vacías al final. Pandas va a leer esto y nos va a crear columnas inútiles llenas de valores nulos (NaN) llamadas Unnamed.

Tipos de datos mezclados: La columna de duración dice "90 min" o "2 Seasons". Al incluir letras, Pandas no la va a tratar como un número matemático, sino como un texto (string).

```python
import pandas as pd

df_netflix = pd.read_csv("netflix_titles.csv", encoding="latin-1")

print("Las columnas del dataset son:")
print(df_netflix.columns.tolist())

df_netflix.head()
```

```python
df_netflix.describe()
```

Si, faltan muchas columnas haciendo describe().! Al ver el resultado, notamos que la única columna original que aparece es release_year (año de lanzamiento), acompañada por un montón de columnas vacías generadas por error (Unnamed: 12, Unnamed: 13, etc.). No están los títulos, ni los directores, ni el país.
Esto ocurre porque el método describe() de Pandas, por defecto, solo calcula estadísticas sobre variables cuantitativas (numéricas). Todas las demás columnas del dataset son detectadas como variables cualitativas (texto), y Pandas las excluye automáticamente porque es imposible calcularle un "promedio" o un "máximo" matemático a un bloque de texto.

show_id: Cualitativa nominal. Es simplemente un identificador alfanumérico único para cada obra.

type: Cualitativa nominal (binaria). Indica la categoría del contenido, con solo dos opciones: "Movie" o "TV Show".

title: Cualitativa nominal. Texto que representa el nombre de la película o serie.

director, cast (elenco), country (país): Cualitativas nominales. Son variables de texto sin un orden jerárquico. (Ojo: a veces incluyen varios valores separados por comas).

date_added: Cuantitativa (formato fecha) o Cualitativa Ordinal (si la pensamos como una cronología temporal). Indica cuándo se subió a la plataforma.

release_year: Cuantitativa discreta. Año original en que se estrenó la obra (número entero sin decimales).  

rating: Cualitativa ordinal. Es la clasificación por edades (TV-Y, PG-13, TV-MA). Existe una relación de orden lógico desde apto para todo público hasta contenido para adultos.  

duration: Cuantitativa discreta (representa minutos o temporadas), pero como está mezclada con texto ("90 min", "2 Seasons"), Pandas la detecta temporalmente como Cualitativa nominal (texto).

listed_in: Cualitativa nominal. Los géneros de la obra.

description: Cualitativa nominal (texto no estructurado).

Columnas Unnamed (12 a 25): Son columnas basura llenas de valores nulos (NaN), generadas automáticamente porque Pandas leyó comas accidentales al final de las filas.

```python
#i)Cuantos direactores distintos hay
cantidad_directores = df_netflix['director'].nunique()
print(f"i)Hay {cantidad_directores} directores distintos registrados")
```

```python
#ii)Quien es el director con mas peliculas
solo_peliculas = df_netflix[df_netflix['type'] == 'Movie']
director_top = solo_peliculas['director'].value_counts().index[0]
cantidad_top = solo_peliculas['director'].value_counts().iloc[0]
print(f"ii) El director con más películas es {director_top} (con {cantidad_top} películas).")
```

```python
#iii)Cuantas series hay?y cuantos programas de tv?
conteo_tipos = df_netflix['type'].value_counts()
print(f"iii) En la base de datos hay:\n{conteo_tipos}")
```

```python
#iv)Cual es la peli mas vieja del catalogo?y la serie?
peli_mas_vieja = solo_peliculas.sort_values(by='release_year').iloc[0]
solo_series = df_netflix[df_netflix['type'] == 'TV Show']
serie_mas_vieja = solo_series.sort_values(by='release_year').iloc[0]

print(f"iv) Película más vieja: '{peli_mas_vieja['title']}' (del año {peli_mas_vieja['release_year']})")
print(f"iv) Serie más vieja: '{serie_mas_vieja['title']}' (del año {serie_mas_vieja['release_year']})")
```

```python
#v)Dado un pais, determinar cuantas peliculas de ese pais hay
pais_elegido = "Argentina"
pelis_del_pais = solo_peliculas[solo_peliculas['country'].fillna('').str.contains(pais_elegido)]
cantidad_pais = len(pelis_del_pais)
print(f"v) Hay {cantidad_pais} películas producidas (o co-producidas) en {pais_elegido}.")
```

Análisis de los tipos de datos en catalogo_streaming_argentina_2024:

title (Título): Cualitativa nominal. Representa el nombre del contenido.

platform (Plataforma): Cualitativa nominal. Indica dónde está disponible (Netflix, Max, Disney, etc.).

type (Tipo): Cualitativa nominal (binaria). Clasifica la obra en MOVIE o TV SHOW.

imdbVotes (Votos de IMDB): Cuantitativa discreta. Es el conteo de cuánta gente votó en la plataforma IMDB (aunque Pandas a veces le agregue un .0 decimal por defecto).

imdbScore (Puntaje de IMDB): Cuantitativa continua. Promedio de calificación (0.0 a 10.0).

tmdbPopularity (Popularidad TMDB): Cuantitativa continua. Índice matemático de relevancia.

tmdbScore (Puntaje de TMDB): Cuantitativa continua. Promedio de calificación en TMDB.

runtime (Duración): Cuantitativa discreta. Medida en minutos.

genres (Géneros): Cualitativa nominal. Categorías temáticas.

director: Cualitativa nominal. Quien dirigió la obra.

year_range (Rango de años): Cualitativa ordinal. Expresa en qué momento salió el título, a veces agrupando en rangos (ej: 1990_2002), lo cual indica una época.

¿Cuál es la plataforma dominante (con más títulos) en el mercado argentino actual?

¿Cuál es la obra con mayor duración (runtime) para armar una maratón extrema?

¿Cuál es la mejor película estrictamente del catálogo de Max (según IMDb)?

```python
# 1. Plataforma con más contenido en Argentina
conteo_plataformas = df_streaming['platform'].value_counts().index[0]
conteo_plataformas_count = df_streaming['platform'].value_counts().iloc[0]
print(f"1) Cantidad de contenido por plataforma:\n{conteo_plataformas} con {conteo_plataformas_count} contenidos\n")
```

```python
# 2. Obra con mayor duración (runtime)
# Ordenamos de mayor a menor (ascending=False) y nos quedamos con el primer registro
obra_mas_larga = df_streaming.sort_values(by='runtime', ascending=False).iloc[0]
print(f"2) La obra más larga es '{obra_mas_larga['title']}' y dura {obra_mas_larga['runtime']} minutos.\n")
```

```python
# 3. Mejor película en Max
# Filtramos que sea Max Y que sea Película
peliculas_max = df_streaming[(df_streaming['platform'] == 'Max') & (df_streaming['type'] == 'MOVIE')]
mejor_peli_max = peliculas_max.sort_values(by='imdbScore', ascending=False).iloc[0]
print(f"3) La mejor película en Max es '{mejor_peli_max['title']}' con un puntaje de {mejor_peli_max['imdbScore']}.\n")
```

```python
# 4. Cantidad de Documentales
# Usamos el truco ninja del .str.contains() y .fillna('') para evitar errores con vacíos
documentales = df_streaming[df_streaming['genres'].fillna('').str.contains('Documental')]
print(f"4) Hay {len(documentales)} títulos etiquetados como 'Documental' en total.\n")
```

```python
# 5. Director estrella en Netflix
# Filtramos solo Netflix, sacamos la columna directores y armamos el conteo
titulos_netflix = df_streaming[df_streaming['platform'] == 'Netflix']
mejor_director = titulos_netflix['director'].value_counts().index[0]
mejor_director_cantidad = titulos_netflix['director'].value_counts().iloc[0]
print(f"5) El director con más títulos en Netflix es {mejor_director} (con {mejor_director_cantidad} obras).")
```
