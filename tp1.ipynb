```python
def es_palindromo(palabra):
    palabra = palabra.lower()
    return palabra == palabra[::-1]

print("¿'neuquen' es un palíndromo?:", es_palindromo("neuquen"))
print("¿'reconocer' es un palíndromo?:", es_palindromo("reconocer"))
print("¿'python' es un palíndromo?:", es_palindromo("python"))
```

```python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def calcular_area(self):
        return self.ancho * self.alto

    def calcular_perimetro(self):
        return 2 * (self.ancho + self.alto)

mi_rectangulo = Rectangulo(4, 5)
print(f"El área del rectángulo es: {mi_rectangulo.calcular_area()}")
print(f"El perímetro del rectángulo es: {mi_rectangulo.calcular_perimetro()}")
```

```python
class Cuadrado(Rectangulo):
    def __init__(self, lado):
        super().__init__(lado, lado)

mi_cuadrado = Cuadrado(5)
print(f"El área del cuadrado es: {mi_cuadrado.calcular_area()}")
print(f"El perímetro del cuadrado es: {mi_cuadrado.calcular_perimetro()}")
```
